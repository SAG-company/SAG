# 🏠 app.py — 홈(메인) 페이지

Streamlit 멀티페이지 앱의 **진입점**. 실행: `streamlit run app.py`.
`pages/` 폴더의 파일들이 자동으로 좌측/상단 메뉴가 되고, 이 파일이 첫 화면(Home)이다.

> ⚠️ 이 노트북은 **문서용**이다. Streamlit 런타임 밖(주피터)에서는 실제 화면이 렌더되지 않는다.

## 0. 라이브러리 임포트
- `base64`: 로컬 이미지를 HTML에 인라인으로 박기 위한 인코딩
- `ui`: 공통 CSS/내비게이션 함수 (프로젝트 자체 모듈)

```python
import base64
from pathlib import Path

import streamlit as st

from ui import inject_global_css, render_nav
```

## 1. 페이지 기본 설정
브라우저 탭 제목·아이콘, 와이드 레이아웃, 사이드바 접힘 상태를 지정한다. (st 명령 중 **가장 먼저** 호출되어야 함)

```python
st.set_page_config(
    page_title="AI 수의사",
    page_icon="🐾",
    layout="wide",
    initial_sidebar_state="collapsed",
)
```

## 2. 이미지 → base64 변환 헬퍼
로컬 이미지 파일을 base64 문자열로 바꿔 `<img src="data:image/png;base64,...">` 형태로 HTML에 직접 삽입한다.
파일이 없으면 빈 문자열을 반환해 화면이 깨지지 않게 한다.

```python
def image_to_base64(path: str) -> str:
    image_path = Path(path)

    if not image_path.exists():
        return ""

    with open(image_path, "rb") as file:
        return base64.b64encode(file.read()).decode()
```

## 3. 전역 CSS 주입 + 상단 내비게이션
`ui.py`의 함수로 사이트 공통 스타일을 넣고, 현재 페이지를 'Home'으로 표시한 내비바를 그린다.

```python
inject_global_css()
render_nav("Home")
```

## 4. 히어로(메인 비주얼) 배경 준비
`assets/app.png`가 있으면 어둑한 오버레이를 얹은 배경 이미지로, 없으면 네이비 그라데이션으로 대체한다.

```python
hero_base64 = image_to_base64("assets/app.png")

if hero_base64:
    hero_background = (
        "linear-gradient(rgba(0,0,0,0.05), rgba(0,0,0,0.08)), "
        f"url('data:image/png;base64,{hero_base64}')"
    )
else:
    hero_background = "linear-gradient(135deg, #08111F, #13284B)"
```
