```python
import sys
import torch

print(f"1. 파이썬 버전: {sys.version.split()[0]}")
print(f"2. PyTorch 버전: {torch.__version__}")

# GPU 가속(CUDA) 가능 여부 확인
cuda_available = torch.cuda.is_available()
print(f"3. GPU 가속 가능 여부: {cuda_available}")

if cuda_available:
    print(f"4. 사용 가능한 GPU 개수: {torch.cuda.device_count()}")
    print(f"5. 현재 선택된 GPU 번호: {torch.cuda.current_device()}")
    print(f"6. GPU 모델명: {torch.cuda.get_device_name(0)}")

    # 실제 GPU 연산 테스트 (행렬 곱셈)
    print("\n--- 실제 GPU 연산 테스트 ---")
    try:
        x = torch.rand(3, 3).cuda()
        y = torch.rand(3, 3).cuda()
        z = torch.matmul(x, y)
        print("정상적으로 GPU 장치에서 연산이 완료되었습니다!")
        print(f"연산 결과 위치: {z.device}")
    except Exception as e:
        print(f"GPU 연산 중 에러 발생: {e}")
else:
    print("\n🚨 경고: 현재 CPU 모드로 동작 중입니다. 커널 설정을 확인하세요.")
```

```python
# 하드웨어 장치 지정 (GPU가 있으면 cuda, 없으면 cpu)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"현재 사용할 장치: {device}")

# 예시: 텐서를 생성하고 GPU 메모리로 이동
x = torch.randn(100, 100).to(device)
```
